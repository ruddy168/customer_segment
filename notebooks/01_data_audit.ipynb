```python
# Defining each cleaning condition

missing_customer = df["Customer ID"].isna()
duplicate = df.duplicated()
non_positive_price = df["Price"] <= 0
cancelled_invoice = df["Invoice"].astype(str).str.startswith("C")
negative_quantity = df["Quantity"] < 0

print("Original rows:", len(df))

print("\nRows affected by each condition:")
print("Missing Customer ID:", missing_customer.sum())
print("Exact duplicates:", duplicate.sum())
print("Price <= 0:", non_positive_price.sum())
print("Cancelled invoices:", cancelled_invoice.sum())
print("Negative quantity:", negative_quantity.sum())

# Proposed clean dataset
clean_mask = (
    ~missing_customer
    & ~duplicate
    & ~non_positive_price
    & ~cancelled_invoice
    & ~negative_quantity
)

print("\nRows remaining after ALL rules:", clean_mask.sum())
print("Rows removed:", (~clean_mask).sum())
print(
    "Percentage retained:",
    round(clean_mask.mean() * 100, 2),
    "%"
)
```

```python
# Creating the clean customer-sales dataset

df_clean = df.loc[clean_mask].copy()

print("Clean dataset shape:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isna().sum())

print("\nRemaining negative quantities:",
      (df_clean["Quantity"] < 0).sum())

print("Remaining non-positive prices:",
      (df_clean["Price"] <= 0).sum())

print("Remaining cancelled invoices:",
      df_clean["Invoice"].astype(str).str.startswith("C").sum())

print("Remaining duplicate rows:",
      df_clean.duplicated().sum())
```

```python
# Saving cleaned transaction data for downstream analysis

CLEAN_FILE_PATH = "../data/clean_transactions.csv"

df_clean.to_csv(CLEAN_FILE_PATH, index=False)

print(f"Clean dataset saved to: {CLEAN_FILE_PATH}")
print(f"Rows: {len(df_clean):,}")
```
